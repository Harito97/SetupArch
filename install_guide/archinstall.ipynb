# Step 1: Prepare

## USB boot (or online boot)
Have a boot USB about Arch Linux (should have latest Arch ISO version). More info in [create boot usb from iso](https://linuxconfig.org/how-to-make-a-bootable-usb-from-an-iso-in-linux). Simple way use:

```bash
sudo dd bs=4M if=/path/to/file.iso of=/dev/sdX status=progress oflag=sync
# /dev/sdX should be the mount of the usb partition
```
  
  
## Internet
A connection with Internet (the easiest way is use Ethernet, otherwise use iwctl - default with Arch ISO boot). 

More info in [connect wifi network with iwctl](https://unix.stackexchange.com/questions/477488/connect-to-wifi-from-command-line-on-linux-systems-through-the-iwd-wireless-dae) or [arch wiki](https://wiki.archlinux.org/title/iwd). 

If iwctl have bug, more info in [can't connect to wifi in iwd](https://www.reddit.com/r/archlinux/comments/15ryoi8/cant_connect_to_wifi_in_iwd/).


# Step 2: Start install

```markdown
- Command: archinstall
- Config:
    |___Keyboard, layout and locale     Default
    |___Mirrors                         "vietnam, global"
    |___Create partition, set mount points (most importance) 
    |               |___Use Manual Partition
    |               |___Select partition and mark mount /boot and / - remember mark / partition can be "modifier"
    |               |___We also can create the /home partition independence with the / partition
    |               |___Don't mark modifier for partition that you don't want to be format - that should "existing".
    |                   /boot partition don't need to be mark "modifier"
    |
    |___Bootloader                      Grub (default)
    |___swap                            True
    |___hostname                        change it to something you like (Eg: archlinux)
    |___root password                   set a root password
    |___user account                    create an account with id and password to login to arch system (mark it root permission)
    |___Profile (Desktop GUI)           Minimal (I want to setup later)
    |___Audio*                          Pirewire
    |___Kernels                         linux (I just want linux)
    |___Additional package              sudo, vim, nano, git, iwd (for connect by wifi - or no need if you have a Ethernet) 
    |___Network configuration           Network Manager (lazy to setup network so chose this)
    |___Timezone                        Ho_Chi_Minh (UTC+7)
    |___Others                          Make default
```

# Step 3: Finish

Wait then go out live mode USB, then:
```bash
reboot
```

# Auto mount partition for work

Use /etc/fstab:
```bash
sudo pacman -S ntfs-3g # for ntfs partition
lsblk -f    # to show the uuid of partitions
sudo nano /etc/fstab
```

Make the content like this:
```markdown
# Static information about the filesystems.
# See fstab(5) for details.

# <file system> <dir> <type> <options> <dump> <pass>

## Mount root
# /dev/nvme0n1p6
UUID=488c47bb-ef77-4e88-8b02-2e030fc1d618       /               ext4            rw,relatime     0 1

## Mount boot
# /dev/nvme0n1p1 LABEL=SYSTEM_DRV
UUID=8CC0-E4F8          /boot           vfat            rw,relatime,fmask=0022,dmask=0022,codepage=437,iocharset=ascii,shortname=mixed,utf8,errors=remount-ro   0 2

## Mount data-project dir
UUID=BEEA9DA6EA9D5C09   /Data           ntfs-3g         rw,relatime     0 2
UUID=00CCC158CCC1491C   /Win11          ntfs-3g         rw,relatime     0 2
```

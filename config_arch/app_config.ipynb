# Screenshot in ArchLinux with WM Hyprland

(Usually this function was be deploy by default)

Hyprland là một Window Manager (WM) dựa trên Wayland, và để chụp ảnh màn hình trong môi trường này, bạn có thể sử dụng các công cụ hỗ trợ Wayland như `grim` kết hợp với `slurp` để chọn vùng chụp màn hình.

Dưới đây là các bước chi tiết:

## 1. Cài đặt các công cụ cần thiết

Đầu tiên, bạn cần cài đặt `grim` và `slurp`. Bạn có thể cài đặt chúng từ kho AUR (Arch User Repository) hoặc từ các gói chính thức nếu có.

Cài đặt từ AUR (nếu cần):

```bash
sudo pacman -Syu
sudo pacman -S grim slurp
```

## 2. Chụp ảnh màn hình toàn bộ màn hình

Bạn có thể chụp ảnh toàn bộ màn hình bằng lệnh:

```bash
grim screenshot.png
```

Lệnh này sẽ lưu ảnh chụp màn hình vào tệp `screenshot.png` trong thư mục hiện tại.

## 3. Chụp ảnh màn hình một vùng cụ thể

Sử dụng `slurp` để chọn vùng chụp màn hình:

```bash
grim -g "$(slurp)" screenshot.png
```

Lệnh này sẽ cho phép bạn chọn một vùng trên màn hình và lưu ảnh chụp của vùng đó vào tệp `screenshot.png`.

## 4. Tích hợp với phím tắt

### Method 1.

Bạn có thể tích hợp các lệnh chụp ảnh màn hình vào các phím tắt để thuận tiện hơn. Tùy vào cách bạn cấu hình phím tắt trong Hyprland, bạn có thể thêm các lệnh tương tự vào tập tin cấu hình của Hyprland.

Ví dụ, nếu bạn sử dụng `sxhkd` để quản lý phím tắt, bạn có thể thêm vào tập tin cấu hình của `sxhkd` (thường là `~/.config/sxhkd/sxhkdrc`):

```plaintext
# Chụp toàn bộ màn hình
Print
    grim ~/Pictures/screenshot_$(date +%F_%T).png

# Chụp một vùng cụ thể
Shift+Print
    grim -g "$(slurp)" ~/Pictures/screenshot_$(date +%F_%T).png
```

Sau đó, khởi động lại `sxhkd` để áp dụng các thay đổi.

### Method 2. 

Tập tin cấu hình của Hyprland thường nằm trong `~/.config/hypr/hyprland.conf`. Mở tập tin này bằng trình soạn thảo văn bản bạn yêu thích, chẳng hạn như `nano` hoặc `vim`:

```bash
nano ~/.config/hypr/hyprland.conf
```

Thêm các dòng sau vào tập tin cấu hình để thiết lập các phím tắt cho việc chụp ảnh màn hình:

```plaintext
# Screenshort shortcut
# Chụp toàn bộ màn hình
bind = Print, exec, ~/.config/hypr/screenshot.sh full

# Chụp một vùng cụ thể
bind = Shift+Print, exec, ~/.config/hypr/screenshot.sh region
```

Hyprland có thể không trực tiếp hỗ trợ việc sử dụng exec để chạy các lệnh phức tạp như bạn mong muốn. Để giải quyết vấn đề này, bạn có thể sử dụng một tập lệnh shell (shell script) để thực hiện các lệnh chụp ảnh màn hình và sau đó gọi tập lệnh này từ cấu hình Hyprland.

```bash
nano ~/.config/hypr/screenshot.sh

#!/bin/bash

case $1 in
  full)
    grim ~/Pictures/screenshot_$(date +%F_%T).png
    ;;
  region)
    grim -g "$(slurp)" ~/Pictures/screenshot_$(date +%F_%T).png
    ;;
  *)
    echo "Usage: $0 {full|region}"
    exit 1
    ;;
esac

chmod +x ~/.config/hypr/screenshot.sh
```

## 5. Xem lại ảnh chụp màn hình

Bạn có thể xem lại các ảnh chụp màn hình trong thư mục đã chỉ định (`~/Pictures` trong ví dụ trên) bằng cách sử dụng bất kỳ trình xem ảnh nào bạn thích, chẳng hạn như `feh`:

```bash
feh ~/Pictures/screenshot_$(date +%F_%T).png
```

Với các bước trên, bạn có thể dễ dàng chụp ảnh màn hình trên Arch Linux với Hyprland. Nếu bạn gặp phải vấn đề nào, hãy cho tôi biết để tôi có thể hỗ trợ thêm.
